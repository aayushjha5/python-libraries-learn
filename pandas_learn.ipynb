 Pandas is a  open source data analysis library written in python
 It leverages the power and speed of numpy to make data analysis and preprocessing easy for data scientists.
 it provides rich and highly robust data operations.

 Pandas has 2 types of DS :  

  a) Series    - It is a one dimensional aray with indexes
               - It stores a single column or row of data in a Dataframe
  b) Dataframe - It's tabular spreadsheet like structure representing rows
                 each of which contains one or multiple columns

 A 1D array(labeled) capable of holding any type of data - Series
 A 2D data (labeled) structure with columns of potentially different
    types of data - Dataframe

```python
# creating a new df

df = pd.DataFrame({
    "name" : ['Alfred', 'Batman', 'Alfred'],
    "toy" : [np.nan, 'Batmobile', 'Bullwhip'],
    "born" : [pd.NaT , pd.Timestamp("1940-04-25"), pd.NaT]
})
```

```python
df
```

```python
# df.dropna()       # it will drop wherever Na is mentioned
```

```python
# df.dropna(how='all', axis=1) # it will remove only if all rows for the col is Na , given axis=1 for column
```

```python
# to remove duplicates

df.drop_duplicates(subset=['name'])     # it will remove the duplicates from name column
```

```python
# if we want 1st occurence and want to remove others,
df.drop_duplicates(subset=['name'], keep='first') 
```

```python
# if we want last occurence and want to remove others,
df.drop_duplicates(subset=['name'], keep='last') 
```

```python
#  inspect a dataset

df.shape
```

```python
df.info()      # will tell the class, rangeIndex, data columns , DT, memory usages
```

```python
# to get unique values count
df['toy'].value_counts(dropna=False)
```

```python
# it will tell whether null is  not there or is it
# it is reverse of isnull()
df.notnull()
```
