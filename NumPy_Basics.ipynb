# NumPy Basics

why use NumPy ?

NumPy is a powerful library for numerical computing in Python. Here are some reasons why you might want to use NumPy:

1. NumPy provides efficient storage
2. It also provides better ways of handling data(functions) for processing
3. It is fast
4. It is easy to learn
5. NumPy uses relatively less memory to store data


```python
import numpy as np
```

np array creation methods

1. Conversion from other Python source(tuple,list)
2. Intrinsic numpy array creation object(arange, ones, zeros)
3. Reading arrays from disk, either from standard or custom formats
4. Creating arrays from raw bytes through use of strings or buffers
5. Use of special library function ex - random
6. Replicating, joining, or mutating existing arrays

Axis in numpy

in 1D , [1,2,3,4,5] -> total axis = 1 [Axis0]
in 2D, [[1,2,3]
        [4,5,6]
        [7,8,9]] -> total axis = 2 [Axis0,Axis1]

axis0 will be row(down) , axis1 will be column(right)
....................
in nD , [         ] -> total axis = n [Axis0, ... , Axis(n-1), Axisn]
