```python
nltk.download('punkt_tab') # word_tokenize relies on sentence tokenization under the hood, and the Punkt resources provide the sentence-boundary model that helps split text into sentences before splitting those sentences into word tokens; without these resources, word_tokenize may fail to load or perform poorly on boundaries and abbreviations.
```

```python
from nltk.probability import FreqDist
fdist = FreqDist() # A statistical representation that shows how often each unique item (word, token, etc.) appears in a given text or corpus.
```

```python
for word in Alex_tokens:
    fdist[word.lower()]+=1
fdist  # shows which tokens appears how many times in the token list
```

```python
# Stemming Operations using PorterStemmer
from nltk.stem import PorterStemmer
pst = PorterStemmer()
```

```python
pst.stem("having")   # stem the word having
```

```python
# for a list of words
words_to_stem = ['give','giving','gave','given']
for word in words_to_stem:
    print(word+ ":" + pst.stem(word))
```

```python
# Lancaster Stemmer
from nltk.stem import LancasterStemmer
lst = LancasterStemmer()
for word in words_to_stem:
    print(word+ ":" + lst.stem(word))
```

```python
# We can conclude that LancasterStemmer is more aggressive than PorterStemmer as it stems the words to a greater extent.
```

```python
# Snowball Stemmer
from nltk.stem import SnowballStemmer
sbst = SnowballStemmer(language='english')
for word in words_to_stem:
    print(word+ ":" + sbst.stem(word))
```

```python
# POS Tagging
nltk.download('averaged_perceptron_tagger_eng')

sent =  "Rahul is a natural when it come to bullshitting"
# first tokenize and then do POS
sent_tokens = word_tokenize(sent)
```

```python
for token in sent_tokens:
    print(nltk.pos_tag([token])) # it will show the POS tag for each token
```

```python
# another example
sent2 =  "Oh, I am in for now, but we are out later"
sent2_tokens = word_tokenize(sent2)
for token in sent2_tokens:
    print(nltk.pos_tag([token]))
```

```python
# Named Entity Recognition (NER)
from nltk import ne_chunk,Maxent_NE_Chunker
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
```

```python
NE_sent = "Donald Trump called Narendra Modi 4 times from the WHITE HOUSE"

# to do NER , first tokenize and do POS tagging
NE_tokens = word_tokenize(NE_sent)
NE_tags = nltk.pos_tag(NE_tokens)

```

```python
NE_NER = ne_chunk(NE_tags)
print(NE_NER)
```

```python
# Chunking Example
new = "The big cat is chasing the little mouse who was after fresh cheese"
new_tokens = word_tokenize(new)
new_tags = nltk.pos_tag(new_tokens)

```

```python
# define the grammer
grammer_np = r"NP: {<DT>?<JJ>*<NN>}"  # NP is noun phrase, DT is determiner, JJ is adjective, NN is noun
```

```python
chunk_parser = nltk.RegexpParser(grammer_np) # it will create a chunk parser object with the defined grammer which will be used to parse the POS tagged tokens
```

```python
chunk_result = chunk_parser.parse(new_tags)
chunk_result
```
